```python
df = pd.read_csv(
    "/kaggle/input/competitions/finclub-open-project-26/dataset.csv"
)
```

```python
df['datetime'] = pd.to_datetime(
    df['datetime'],
    format='%d-%m-%Y %H:%M'
)

df = df.sort_values('datetime')
```

```python
import re

option_cols = [
    c for c in df.columns
    if c not in ['datetime', 'underlying_price']
]

strike_map = {}

for col in option_cols:

    strike = int(
        re.search(r'(\d+)(CE|PE)$', col).group(1)
    )

    option_type = re.search(
        r'(CE|PE)$',
        col
    ).group(1)

    strike_map[col] = {
        'strike': strike,
        'type': option_type
    }
```

```python
filled = df.copy()

for col in option_cols:

    filled[col] = (
        filled[col]
        .interpolate(method='linear')
        .ffill()
        .bfill()
    )
```

```python
ce_cols = sorted(
    [c for c in option_cols if c.endswith('CE')],
    key=lambda x: strike_map[x]['strike']
)

pe_cols = sorted(
    [c for c in option_cols if c.endswith('PE')],
    key=lambda x: strike_map[x]['strike']
)
```

```python
for idx in df.index:

    ce_vals = pd.to_numeric(
    df.loc[idx, ce_cols],
    errors='coerce'
)

    df.loc[idx, ce_cols] = (
        ce_vals.interpolate()
               .ffill()
               .bfill()
    )

    pe_vals = pd.to_numeric(
    df.loc[idx, pe_cols],
    errors='coerce'
)

    df.loc[idx, pe_cols] = (
        pe_vals.interpolate()
               .ffill()
               .bfill()
    )
```

```python
strike_fill = df.copy()

strike_fill[ce_cols] = (
    strike_fill[ce_cols]
    .apply(pd.to_numeric, errors='coerce')
    .interpolate(axis=1)
    .ffill(axis=1)
    .bfill(axis=1)
)

strike_fill[pe_cols] = (
    strike_fill[pe_cols]
    .apply(pd.to_numeric, errors='coerce')
    .interpolate(axis=1)
    .ffill(axis=1)
    .bfill(axis=1)
)
```

```python
print(strike_fill.isna().sum().sum())
```

```python
filled_dataset = strike_fill.copy()

filled_dataset.to_csv(
    "filled_dataset.csv",
    index=False
)
```

```python
import pandas as pd

ORIGINAL_DATASET_PATH = "/kaggle/input/competitions/finclub-open-project-26/dataset.csv"

SEPARATOR = "||"

def generate_solution(filled_path, output_path="submission.csv"):

    original = pd.read_csv(ORIGINAL_DATASET_PATH)
    filled = pd.read_csv(filled_path)

    feature_cols = [
        c for c in original.columns
        if c != "datetime"
    ]

    rows = []

    for col in feature_cols:

        was_missing = original[col].isna()

        for idx in original.index[was_missing]:

            dt = original.loc[idx, "datetime"]

            uid = f"{dt}{SEPARATOR}{col}"

            val = filled.loc[idx, col]

            rows.append({
                "id": uid,
                "value": val
            })

    solution = pd.DataFrame(
        rows,
        columns=["id", "value"]
    )

    solution = (
        solution
        .sort_values("id")
        .reset_index(drop=True)
    )

    solution.to_csv(
        output_path,
        index=False
    )

    print(
        f"Saved {len(solution)} rows"
    )

generate_solution(
    "filled_dataset.csv",
    "submission.csv"
)
```
